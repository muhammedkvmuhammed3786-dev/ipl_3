```python
df=pd.read_sql("""select phase , count(*) as balls from v_ball group by phase;""",con)
df.plot(kind="bar",
        x="phase",
        y="balls",
        xlabel="phase",
        ylabel="total no:of balls",
        title="total balls in each phases")
```

```python
df=pd.read_sql(""" 
 --give the list of highest runs scored in ipl history
 --max 5
select batter,sum(batter_runs) as total_runs
from v_ball
group by batter
order by total_runs desc
limit 5;""",con)
df.plot(kind="bar",
        x="batter",
        y="total_runs",
        xlabel="batters",
        ylabel="total runs",
        title="highest runs scored in ipl history")
```

```python

df=pd.read_sql("""
-- list of highest six scores in ipl history

  select batter,
         sum (case when batter_runs=6
              then 1
              else 0
              end)as total_sixes
from v_ball
group by batter
order by total_sixes desc
limit 10;""",con)
df.plot(kind="bar",
        x="batter",
        y="total_sixes",
        color=['darkviolet', 'indigo', 'blue', 'green', 'yellow', 'orange', 'red'],
        legend=False,
        xlabel="batters",
        ylabel="total sixes",
        title="highest six scores in ipl history")
```
